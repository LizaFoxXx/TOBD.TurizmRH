## Постановка бизнес-задачи

### Бизнес-проблема

Туристический спрос в России распределён неравномерно по территории и времени. Региональным органам управления туризмом и организациям индустрии гостеприимства необходимо понимать, в какие месяцы инфраструктура перегружена, где возникает продолжительный низкий сезон и какие территории имеют резерв для развития круглогодичного предложения.

Отсутствие единого сопоставимого профиля приводит к тому, что решения о продвижении межсезонья, развитии инфраструктуры и распределении ресурсов могут приниматься без количественного обоснования. При этом одинаковая сезонная амплитуда может иметь разные причины: пляжную специализацию, зимний отдых, событийный календарь, климат или недостаточную диверсификацию предложения.

### Заинтересованные стороны

- региональные органы исполнительной власти в сфере туризма;
- туристско-информационные центры;
- коллективные средства размещения;
- туроператоры, организаторы мероприятий и транспортные организации;
- аналитики внутреннего туризма и регионального развития.

### Бизнес-цель

**Повысить обоснованность планирования сезонного продвижения и использования туристической инфраструктуры субъектов Российской Федерации за счёт аналитического прототипа, который выявляет пики и спады спроса, сравнивает регионы по сезонности, оценивает резерв низкого сезона и формирует адресные рекомендации для разных типов территорий.**

Цель состоит не в механическом увеличении турпотока любой ценой, а в более равномерном использовании существующей инфраструктуры при сохранении годового числа ночёвок.

## Измеримые критерии достижения цели

1. **Полнота аналитического профиля:** для региона доступно не менее 36 помесячных наблюдений и не более 5% пропусков в ключевых показателях.
2. **Измерение сезонности:** для каждого региона рассчитаны сезонный профиль из 12 месяцев, коэффициент вариации и амплитуда.
3. **Определение экстремумов:** для каждого региона определены три пиковых и три низких месяца.
4. **Резерв низкого сезона:** рассчитан недобор ночёвок относительно среднемесячного уровня в абсолютном и относительном выражении.
5. **Приоритизация территорий:** составлен список не менее 15 регионов с высокой сезонностью и большим резервом низкого сезона.
6. **Региональная типология:** выделено не менее трёх содержательно интерпретируемых профилей спроса.
7. **Сценарный эффект:** при сценарном переносе 7,5% спроса из трёх пиковых месяцев в три низких месяца коэффициент сезонной вариации снижается минимум на 10% при неизменном годовом числе ночёвок как минимум у 70% приоритетных регионов.
8. **Практический результат:** для каждого типа региона сформирована минимум одна рекомендация, связанная с измеримым показателем.

Порог сценария является проектным ориентиром, а не прогнозом гарантированного поведения туристов.

## Формулы

Для региона \(r\) и месяца \(m\):

\[
S_{r,m}=\frac{\overline{N}_{r,m}}{\overline{N}_{r}},
\]

где \(\overline{N}_{r,m}\) — среднее число ночёвок в месяце \(m\), а \(\overline{N}_{r}\) — среднемесячное число ночёвок региона.

\[
CV_r=\frac{\sigma(S_{r,1},\ldots,S_{r,12})}{\mu(S_{r,1},\ldots,S_{r,12})}.
\]

\[
A_r=\frac{\max_m(S_{r,m})}{\min_m(S_{r,m})}.
\]

\[
R_r=\sum_{m=1}^{12}\max(\overline{N}_{r}-\overline{N}_{r,m},0).
\]

Сценарное снижение сезонности:

\[
E_r=\frac{CV_r^{before}-CV_r^{scenario}}{CV_r^{before}}\cdot100\%.
\]

### Научная цель

Выявить устойчивые сезонные и географические закономерности туристического спроса в субъектах Российской Федерации и оценить связь уровня и неравномерности спроса с климатическими условиями, ёмкостью инфраструктуры размещения и тематическим профилем туристического предложения.

### Объект исследования

**Туристический спрос в субъектах Российской Федерации, наблюдаемый через показатели деятельности коллективных средств размещения.**

### Предмет исследования

**Временные и географические закономерности числа ночёвок, а также их связи с климатическими, инфраструктурными и содержательными характеристиками туристического предложения регионов.**

### Аналитическая единица

\[
\text{субъект РФ}\times\text{год}\times\text{месяц}.
\]

### Границы

- **Территориальные:** субъекты РФ с сопоставимыми официальными данными; федеральные округа и строка «Российская Федерация» не считаются субъектами.
- **Временные:** январь 2023 — декабрь 2025 для расчётов в текущей версии; 2026 год неполный и не включён в базовый сезонный профиль.
- **Содержательные:** учитывается спрос, зарегистрированный коллективными средствами размещения; неформальная аренда, однодневные поездки и проживание у родственников отражаются неполно или не отражаются.
- **Методологические:** выявленные связи не доказывают причинность без экспериментального или квазиэкспериментального дизайна.

### Основная аналитическая гипотеза

**Сезонная неравномерность туристического спроса имеет устойчивую географическую структуру и связана с сочетанием климата, ёмкости средств размещения и тематического профиля туристического предложения региона.**

Частные гипотезы:

- H1: тёплая и солнечная погода положительно связана с летним спросом рекреационных территорий;
- H2: снег и отрицательная температура связаны с зимним спросом регионов с зимним предложением;
- H3: большая ёмкость средств размещения связана с объёмом ночёвок, но не гарантирует низкую сезонность;
- H4: тематически разнообразное предложение связано с меньшей сезонной амплитудой;
- H5: близкие географически регионы могут иметь разные профили из-за специализации и инфраструктуры.

## Настройка окружения

```python
from pathlib import Path
from time import perf_counter
import warnings

import numpy as np
import pandas as pd
import requests
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, Markdown

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", 30)
pd.set_option("display.float_format", lambda x: f"{x:,.3f}")

DATA_DIR = Path("data")
RESULTS_DIR = Path("results")
DATA_DIR.mkdir(exist_ok=True)
RESULTS_DIR.mkdir(exist_ok=True)

ROSSTAT_URL = "https://rosstat.gov.ru/storage/mediabank/SR_mes_sub_08-2026.xlsx"
ROSSTAT_FILE = DATA_DIR / "rosstat_ksr_monthly.xlsx"
YEARS = [2023, 2024, 2025]
MONTH_NAMES = {1:"Январь",2:"Февраль",3:"Март",4:"Апрель",5:"Май",6:"Июнь",7:"Июль",8:"Август",9:"Сентябрь",10:"Октябрь",11:"Ноябрь",12:"Декабрь"}

sns.set_theme(style="whitegrid", font_scale=1.0)
plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["axes.unicode_minus"] = False
print("Каталоги подготовлены:", DATA_DIR.resolve(), RESULTS_DIR.resolve())
```

## Загрузка официального XLSX Росстата

На странице Росстата набор называется **«Основные показатели деятельности коллективных средств размещения по Российской Федерации и субъектам Российской Федерации, месяцы (с 2022 г.)»**. Таблицы внутри файла содержат накопительные значения «январь», «январь–февраль», …, «январь–декабрь». Для получения истинных месячных значений далее берётся первая разность по столбцам.

```python
if not ROSSTAT_FILE.exists():
    t0 = perf_counter()
    response = requests.get(ROSSTAT_URL, timeout=120, verify=False)
    response.raise_for_status()
    ROSSTAT_FILE.write_bytes(response.content)
    print(f"Файл загружен: {len(response.content)/1024:.1f} КБ за {perf_counter()-t0:.2f} с")
else:
    print("Используется локальный файл:", ROSSTAT_FILE)

xls = pd.ExcelFile(ROSSTAT_FILE)
print("Число листов:", len(xls.sheet_names))
print("Первые листы:", xls.sheet_names[:15])
```

## Парсинг ночёвок и размещённых лиц

```python
def parse_cumulative_sheets(metric_prefix: str, years=YEARS) -> pd.DataFrame:
    """Преобразует накопительные таблицы Росстата в длинный помесячный формат."""
    frames = []
    for year in years:
        sheet = f"{metric_prefix}.{year}"
        raw = pd.read_excel(ROSSTAT_FILE, sheet_name=sheet, header=None, skiprows=3)
        raw = raw.iloc[:, :13]
        raw.columns = ["region"] + list(range(1, 13))
        raw = raw[raw["region"].notna()].copy()
        raw["region"] = raw["region"].astype(str).str.strip()
        raw = raw[~raw["region"].str.contains(
            "Российская Федерация|федеральный округ|Без учета|Без учёта",
            case=False, regex=True
        )]
        raw = raw[~raw["region"].str.match(r"^\d")]
        cumulative = raw[list(range(1, 13))].apply(pd.to_numeric, errors="coerce")
        monthly = cumulative.diff(axis=1)
        monthly[1] = cumulative[1]
        for month in range(1, 13):
            frames.append(pd.DataFrame({
                "region": raw["region"],
                "year": year,
                "month": month,
                "value": monthly[month]
            }))
    return pd.concat(frames, ignore_index=True)

nights = parse_cumulative_sheets("2").rename(columns={"value": "nights"})
arrivals = parse_cumulative_sheets("1").rename(columns={"value": "arrivals"})

monthly = nights.merge(arrivals, on=["region", "year", "month"], how="inner")

# В таблице присутствуют одновременно итоги по Архангельской/Тюменской областям
# и строки без автономных округов. Для непересекающегося набора оставляем остаточные строки.
for aggregate_region in ["Архангельская область", "Тюменская область"]:
    monthly = monthly[monthly["region"] != aggregate_region]

monthly["region"] = monthly["region"].replace({
    "Архангельская область (кроме Ненецкого автономного округа)": "Архангельская область",
    "Тюменская область (кроме Ханты-Мансийского автономного округа-Югры и Ямало-Ненецкого автономного округа)": "Тюменская область"
})
monthly["date"] = pd.to_datetime(dict(year=monthly.year, month=monthly.month, day=1))
monthly["month_name"] = monthly["month"].map(MONTH_NAMES)
monthly["avg_stay"] = monthly["nights"] / monthly["arrivals"].replace(0, np.nan)
monthly = monthly.sort_values(["region", "date"]).reset_index(drop=True)

print("Размер витрины:", monthly.shape)
print("Число субъектов:", monthly["region"].nunique())
print("Период:", monthly["date"].min().date(), "—", monthly["date"].max().date())
display(monthly.head())
```

## Контроль качества данных

```python
quality = pd.DataFrame({
    "тип": monthly.dtypes.astype(str),
    "пропусков": monthly.isna().sum(),
    "доля_пропусков": monthly.isna().mean(),
    "уникальных": monthly.nunique()
})
display(quality)

region_quality = monthly.groupby("region").agg(
    observations=("date", "size"),
    missing_nights=("nights", lambda x: x.isna().mean()),
    missing_arrivals=("arrivals", lambda x: x.isna().mean()),
    first_date=("date", "min"),
    last_date=("date", "max")
).reset_index()

eligible = region_quality.query(
    "observations >= 36 and missing_nights <= 0.05 and missing_arrivals <= 0.05"
)
print(f"Критерию полноты соответствуют {len(eligible)} из {len(region_quality)} субъектов.")
display(region_quality.sort_values(["observations", "missing_nights"]).head())

monthly = monthly[monthly["region"].isin(eligible["region"])].copy()
assert not monthly.duplicated(["region", "date"]).any(), "Найдены дубликаты ключа регион–месяц"
assert (monthly[["nights", "arrivals"]] >= 0).all().all(), "Найдены отрицательные значения"

monthly.to_csv(RESULTS_DIR / "monthly_tourism_clean.csv", index=False, encoding="utf-8-sig")
```

### Вывод по качеству

Проверка исключает дубликаты ключа `регион–месяц`, отрицательные значения и регионы с недостаточным числом наблюдений. ## Описательная статистика

```python
year_summary = monthly.groupby("year").agg(
    nights=("nights", "sum"),
    arrivals=("arrivals", "sum")
).reset_index()
year_summary["avg_stay"] = year_summary["nights"] / year_summary["arrivals"]
year_summary["nights_growth"] = year_summary["nights"].pct_change()
year_summary["arrivals_growth"] = year_summary["arrivals"].pct_change()
display(year_summary.style.format({
    "nights": "{:,.0f}", "arrivals": "{:,.0f}", "avg_stay": "{:.2f}",
    "nights_growth": "{:.1%}", "arrivals_growth": "{:.1%}"
}))

top_annual = monthly.groupby("region").agg(
    avg_annual_nights=("nights", lambda x: x.sum()/x.index.to_series().map(lambda _: 1).count()*12),
    total_nights=("nights", "sum"),
    total_arrivals=("arrivals", "sum")
).reset_index()
top_annual["avg_stay"] = top_annual.total_nights / top_annual.total_arrivals
display(top_annual.nlargest(15, "total_nights"))
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
plot_year = year_summary.copy()
plot_year["nights_mln"] = plot_year["nights"] / 1e6
sns.barplot(data=plot_year, x="year", y="nights_mln", color="#2979B9", ax=ax)
ax.set(title="Число ночёвок в КСР по субъектам РФ", xlabel="Год", ylabel="Ночёвки, млн")
for container in ax.containers:
    ax.bar_label(container, fmt="%.1f", padding=3)
plt.tight_layout()
plt.savefig(RESULTS_DIR / "01_nights_by_year.png", dpi=160, bbox_inches="tight")
plt.show()
```

## Расчёт сезонных профилей

```python
profile = monthly.groupby(["region", "month"], as_index=False).agg(
    avg_nights=("nights", "mean"),
    avg_arrivals=("arrivals", "mean")
)
profile["region_mean"] = profile.groupby("region")["avg_nights"].transform("mean")
profile["season_index"] = profile["avg_nights"] / profile["region_mean"]
profile["month_name"] = profile["month"].map(MONTH_NAMES)
profile["gap_to_mean"] = np.maximum(profile["region_mean"] - profile["avg_nights"], 0)

rows = []
for region, g in profile.groupby("region"):
    g = g.sort_values("month")
    peak3 = g.nlargest(3, "avg_nights")["month_name"].tolist()
    low3 = g.nsmallest(3, "avg_nights")["month_name"].tolist()
    rows.append({
        "region": region,
        "cv": g["season_index"].std(ddof=0) / g["season_index"].mean(),
        "amplitude": g["season_index"].max() / g["season_index"].min(),
        "peak_month": int(g.loc[g["season_index"].idxmax(), "month"]),
        "low_month": int(g.loc[g["season_index"].idxmin(), "month"]),
        "peak_3": ", ".join(peak3),
        "low_3": ", ".join(low3),
        "reserve_nights": g["gap_to_mean"].sum(),
        "annual_avg_nights": g["avg_nights"].sum()
    })

metrics = pd.DataFrame(rows)
metrics["reserve_share"] = metrics["reserve_nights"] / metrics["annual_avg_nights"]
metrics["peak_month_name"] = metrics["peak_month"].map(MONTH_NAMES)
metrics["low_month_name"] = metrics["low_month"].map(MONTH_NAMES)

def profile_type(row):
    if row["cv"] < 0.15:
        return "Относительно равномерный"
    if row["peak_month"] in [6, 7, 8]:
        return "Летний пик"
    if row["peak_month"] in [12, 1, 2]:
        return "Зимний пик"
    if row["peak_month"] in [3, 4, 5]:
        return "Весенний пик"
    return "Осенний пик"

metrics["profile_type"] = metrics.apply(profile_type, axis=1)
metrics = metrics.sort_values("cv", ascending=False).reset_index(drop=True)
metrics.to_csv(RESULTS_DIR / "regional_seasonality_metrics.csv", index=False, encoding="utf-8-sig")

display(metrics.head(15).style.format({
    "cv":"{:.1%}", "amplitude":"{:.2f}", "reserve_nights":"{:,.0f}",
    "annual_avg_nights":"{:,.0f}", "reserve_share":"{:.1%}"
}))
```

## Сценарная оценка

Сценарий не является прогнозом. Он отвечает на вопрос: **насколько изменится формальный показатель сезонной неравномерности, если 7,5% среднего спроса трёх наиболее загруженных месяцев перенести в три наименее загруженных месяца без изменения годового объёма?**

```python
TRANSFER_SHARE = 0.075
scenario_rows = []
for region, g in profile.groupby("region"):
    g = g.sort_values("month")
    before = g["avg_nights"].to_numpy(dtype=float)
    order = np.argsort(before)
    low_idx, peak_idx = order[:3], order[-3:]
    after = before.copy()
    transfer = after[peak_idx] * TRANSFER_SHARE
    after[peak_idx] -= transfer
    after[low_idx] += transfer.sum() / 3
    cv_before = np.std(before / before.mean())
    cv_after = np.std(after / after.mean())
    scenario_rows.append({
        "region": region,
        "cv_before": cv_before,
        "cv_after": cv_after,
        "cv_reduction": (cv_before - cv_after) / cv_before if cv_before else 0,
        "annual_before": before.sum(),
        "annual_after": after.sum()
    })

scenario = pd.DataFrame(scenario_rows)
scenario["annual_difference"] = scenario["annual_after"] - scenario["annual_before"]
priority_scenario = priority[["region", "priority_score"]].merge(scenario, on="region")
criterion_share = (priority_scenario["cv_reduction"] >= 0.10).mean()
annual_ok = np.allclose(priority_scenario["annual_before"], priority_scenario["annual_after"])

print(f"Доля приоритетных регионов со снижением CV не менее 10%: {criterion_share:.1%}")
print("Годовой объём сохранён:", annual_ok)
print("Критерий 70% выполнен:", criterion_share >= 0.70)

display(priority_scenario.sort_values("cv_reduction", ascending=False).style.format({
    "priority_score":"{:.3f}", "cv_before":"{:.1%}", "cv_after":"{:.1%}",
    "cv_reduction":"{:.1%}", "annual_before":"{:,.0f}", "annual_after":"{:,.0f}",
    "annual_difference":"{:.6f}"
}))
priority_scenario.to_csv(RESULTS_DIR / "scenario_priority_regions.csv", index=False, encoding="utf-8-sig")
```
